```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample sales data in-memory (no SQL, no DB)
data = [
    {'product': 'Coffee', 'quantity': 100, 'price': 5.0},
    {'product': 'Tea', 'quantity': 150, 'price': 2.5},
    {'product': 'Sugar', 'quantity': 200, 'price': 1.0},
    {'product': 'Milk', 'quantity': 250, 'price': 1.2},
    {'product': 'Coffee', 'quantity': 120, 'price': 5.0},
    {'product': 'Tea', 'quantity': 180, 'price': 2.5},
]

# Create a pandas DataFrame from the data
df = pd.DataFrame(data)

# Calculate revenue per row
df['revenue'] = df['quantity'] * df['price']

# Group by product to get total quantity and revenue
summary = df.groupby('product').agg(
    total_qty=('quantity', 'sum'),
    revenue=('revenue', 'sum')
).reset_index()

# Print the summary
print(summary)

# Plot revenue bar chart
summary.plot(kind='bar', x='product', y='revenue', color='lightgreen', legend=False)
plt.title('Sales Revenue by Product (No SQL)')
plt.xlabel('Product')
plt.ylabel('Revenue ($)')
plt.tight_layout()
plt.savefig('sales_chart_no_sql.png')
plt.show()
```
